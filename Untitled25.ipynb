# Publishing & Sharing Reports in Power BI Assignment

Q1. Explain why publishing and sharing reports is considered more important than just building dashboards
in Power BI Desktop.

Sol :         Data visualizations only provide value when they drive business decisions. Building a report in Power BI Desktop traps the insights on a single user's local machine. Publishing and sharing democratize the data, allowing stakeholders, teams, and decision-makers to view the insights, collaborate, and act upon the information.


Q2. Define Power BI Service and explain its role in comparison to Power BI Desktop in the reporting lifecycle.

Sol :        
 -  Power BI Desktop is a free local application used for authoring (connecting to data, transforming it, modeling it, and creating visuals).
- Power BI Service is the cloud-based SaaS (Software as a Service) platform.
In the reporting lifecycle, Desktop is the development environment where the work is done, while the Service is the deployment and distribution environment where reports are published, shared, automated (via scheduled refresh), and consumed by end-users.

Q3. What does publishing a report mean in Power BI? List any two outcomes that occur after a report is
published.

Sol :   
 -  Outcome 1: The local file is split into two separate artifacts in the cloud: a Dataset (the data model) and a Report (the visuals).
- Outcome 2: The report becomes accessible via a web browser and can now be shared with other users or pinned to web dashboards.

Q4. List and explain the steps involved in publishing a report from Power BI Desktop to Power BI Service in the
correct sequence.

Sol :     
i . Create and save the report in Power BI Desktop.

ii . Sign in to your Power BI account.

iii .Click the Publish button on the Home tab.

iv . Select the target workspace.

v . Confirm publishing.

vi . Power BI uploads the report and dataset to Power BI Service.

vii .Open Power BI Service to verify the report.

viii . Configure refresh schedules and sharing permissions if required.

Q5. Differentiate between Report and Dashboard in Power BI based on structure and usage.

Sol :     
- Report: A multi-page document that provides detailed, interactive visualizations. It is tied to a single underlying dataset. Users can heavily interact with reports using slicers, cross-filtering, and drill-downs to explore data deeply.
- Dashboard: A single-page canvas containing "tiles" (visuals) pinned from multiple different reports and datasets. It is used to provide a high-level, at-a-glance summary for executives or quick monitoring, and offers less interactivity than a full report.

Q6. Explain the difference between the following three sharing methods in Power BI Service:

- Direct Report Sharing

- Workspace Access

- Power BI Apps

Also mention one use case for each.

Sol :     
- Direct Report Sharing : Generating a specific link to share a single report with individual users.
   - Use Case: Quickly sharing an ad-hoc analysis with a specific colleague for a one-off discussion.
- Workspace Access : Granting users roles (like Admin, Member, or Contributor) to a shared collaborative folder where datasets and reports are stored.
   - Use Case: A team of data analysts and developers collaborating to build and test reports together.
- Power BI Apps : Packaging multiple related reports and dashboards into a single, cohesive, read-only application with customized navigation.
   - Use Case: Distributing a finalized, official suite of financial reports to the entire organization or a large department.
      

Q7. Describe the four workspace roles available in Power BI and explain how access control is managed using
these roles.

Sol :  
i . Admin: Has full control. Can add/remove users, delete the workspace, and manage all content.

ii . Member: Can add users with lower permissions, publish apps, share items, and create/edit content.

iii . Contributor: Can create, edit, and delete reports and datasets, but cannot add new users or alter workspace settings.

iv . Viewer: Can only view and interact with reports (read-only access). Cannot edit or share content.

- Access Control: This is managed by applying the principle of least privilege. Developers get Contributor/Member roles, administrators get Admin roles, and standard consumers who only need to look at the data are granted Viewer roles to prevent accidental modifications.

Q8. Why are Power BI Apps considered more secure and scalable than direct sharing?
Explain this from a governance and enterprise usage perspective.

Sol :  
Security Perspective

i . Centralized permission management.

ii . Users receive only published content.

iii . Easier governance and auditing.

iv . Reduces accidental exposure of reports.

 Scalability Perspective

i . One app can be distributed to hundreds or thousands of users.

ii . Updates are published once and automatically available to all users.

iii .Simplifies maintenance and administration.

Enterprise Benefits

- Better governance.
- Consistent user experience.
- Easier compliance and security management.
- Efficient large-scale deployment.

Q9. Explain Row-Level Security (RLS) in Power BI.
How does RLS ensure that different users see different data using the same report?

Sol :  Row-Level Security (RLS) restricts data access for given users at the database row level. Instead of creating a separate report for every region or department, you create one master report and dataset.
- How it works: DAX filters are applied to tables within the Power BI Desktop data model (e.g., Region = "North"). These filters are grouped into "Roles." When published to the Service, users or groups are assigned to these roles. When a user opens the report, Power BI detects their login identity (UPN), applies the assigned RLS filter to the backend queries, and only renders the specific rows of data they are authorized to see on the visuals.

Q10.Assume an organization shares reports without using Apps, workspace roles, or RLS.
Discuss any three risks or issues that may arise related to security, governance, or data misuse, and explain
how Power BI best practices help prevent them.      

Sol :  

- Risk 1: Severe Data Leakage. Without RLS, all users who receive the report link see the exact same data. A regional manager might see another region's sensitive financial data or HR information. Best Practice: Implement RLS based on organizational hierarchy so the same report automatically filters to show only the viewer's permitted data.
- Risk 2: Accidental Deletion and Alteration. If users are just dumped into workspaces without proper role management (e.g., everyone is an Admin or Member), non-technical users might accidentally delete datasets, alter critical DAX measures, or break visuals. Best Practice: Enforce strict Workspace Roles—only developers get Contributor access; consumers get Viewer access or are directed to an App.
- Risk 3: Management Chaos ("Spaghetti Sharing"). Relying on direct sharing creates an unmanageable governance nightmare. When an employee leaves or changes departments, administrators cannot easily trace or revoke all the individual report links that person had access to. Best Practice: Stop direct sharing. Use Power BI Apps distributed to dynamic Entra ID (Azure AD) Security Groups. When an employee switches roles, IT simply updates their AD group, and their Power BI access updates automatically.
